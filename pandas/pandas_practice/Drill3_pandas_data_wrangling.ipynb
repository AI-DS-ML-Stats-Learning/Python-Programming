```python
import pandas as pd

sales = pd.DataFrame({
    'store': ['North', 'North', 'South', 'South'],
    'category': ['Electronics', 'Grocery', 'Electronics', 'Grocery'],
    'Jan': [1200, 800, 950, 1100],
    'Feb': [1500, 750, 1050, 900],
    'Mar': [1100, 900, 1200, 1000]
})
```

##### 1. Wide → Long: Reshape sales from wide (Jan, Feb, Mar as columns) into long format with columns store, category, month, revenue, sorted by ['store', 'category', 'month'].

```python
sales
```

```python
sales.melt(id_vars=['store', 'category'], var_name='month', value_name='revenue').sort_values(by=['store', 'category', 'month'], ascending=[False, False, False])
```

##### 2. Single-Chain Pipeline: Starting from sales, write one single chained expression (no intermediate variables, no inplace=True) that:
        `1. Filters to category == 'Electronics',
        `2. Reshapes Jan, Feb, Mar into month and revenue,
        `3. Adds rev_k (revenue / 1000),
        `4. Adds store_total (total Electronics revenue per store across all 3 months, shown on every row).

```python
sales[sales['category']=='Electronics'].melt(id_vars=['store', 'category'], 
                                             var_name='month', value_name = 'revenue').assign(
                                                 rev_k = lambda sales : sales['revenue']/1000).assign(
                                                     store_total = lambda sales: sales.groupby('store')['revenue'].transform("sum"))
```

```python
sales[sales['category']=='Electronics'].melt(id_vars=['store', 'category'], 
                                             var_name='month', value_name = 'revenue').assign(
                                                 rev_k = lambda sales : sales['revenue']/1000,
                                                     store_total = lambda d: d.groupby('store')['revenue'].transform("sum"))
```
