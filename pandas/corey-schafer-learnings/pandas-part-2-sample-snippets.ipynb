### Filtering a DF - Using conditions to filter rows and columns

```python
df['last'] == 'doe' #this returns a series of true and false values
```

```python
# filt = df['last'] == 'doe'
filt = (df['last'] == 'doe')

df[filt]
```

```python
df[df['last'] == 'doe']
```

```python
df.loc[filt, 'email']
```

```python
# & - and
# | - or
```

```python
filt = (df['last'] == 'doe') & (df['first'] == 'jon')
```

```python
df.loc[filt, 'email']
```

```python
filt = (df['last'] == 'doe') | (df['first'] == 'jon')
```

```python
df.loc[filt, 'email']
```

```python
df.loc[~filt, 'email']
```

### Updating rows and columns - Modifying Data within DataFrames
