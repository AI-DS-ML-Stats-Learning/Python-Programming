#### 33. How to get the dates of yesterday, today and tomorrow? (★☆☆)

````python
# print(np.datetime64('today'))
# print(np.datetime64('today')-1)
# print(np.datetime64('today')+1)

today = np.datetime64('today')
yesterday = today - np.timedelta64(1, 'D')
tomorrow = today + np.timedelta64(1, 'D')

print(yesterday)
print(today)
print(tomorrow)
````
