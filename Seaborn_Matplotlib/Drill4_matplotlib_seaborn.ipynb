```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
eda = pd.DataFrame({
    'category': np.random.choice(['Electronics', 'Grocery', 'Apparel'], size=200),
    'channel': np.random.choice(['Web', 'App'], size=200),
    'order_value': np.concatenate([
        np.random.lognormal(mean=4.2, sigma=0.7, size=195),
        [850, 920, 1100, 1250, 1400]
    ]),
    'items': np.random.randint(1, 8, size=200),
    'discount_pct': np.random.uniform(0, 30, size=200).round(1)
})
```

##### 4 EDA Charts to Build
    ~ 1. Distribution: Plot the distribution of order_value with a smooth density curve, plus a title and x/y axis labels.
    ~ 2. Outliers & Group Split: Plot order_value by category, split by channel, in a chart that clearly shows medians, quartiles, and outliers.
    ~ 3. Category Summary: Plot the mean order_value per category as a bar chart.
    ~ 4. Correlations: Compute the correlation matrix of the numeric columns (order_value, items, discount_pct) and plot it as a color-coded heatmap with the correlation numbers printed inside each cell.

```python
eda
```

```python
# 1. 

plt.figure(figsize=(5,5))
plt.title('Density Curve')
plt.xlabel('Order Value')
plt.ylabel('Density')
sns.histplot(data=eda, kde=True, x='order_value', stat='density')
```

```python
# 2.
plt.title('Order value by Category - Boxplot')
plt.xlabel('Order Value')
plt.ylabel('Category')
sns.boxplot(data=eda, x='order_value', y='category', hue='channel')
```

```python
# 3

sns.barplot(data=eda, x='category', y='order_value', estimator='mean')
```

```python
# 4

corr = eda.corr(numeric_only=True)

sns.heatmap(data=corr, annot = True)
```
